### 🎫 Boarding: the mission

Every passenger who steps off a plane carries a verdict: **satisfied** or **neutral / dissatisfied**.
Our job is to read the passenger's boarding pass (who they are, where they fly, how they rated
13 touchpoints of the journey) and predict the **probability that they leave the plane happy**.

| | |
|---|---|
| 🎯 **Target** | `satisfaction` (1 = satisfied) |
| 📏 **Metric** | ROC AUC: only the *ranking* of passengers matters, not calibrated probabilities |
| 🧳 **Data** | Synthetic data generated from the classic *Airline Passenger Satisfaction* survey |
| 📤 **Output** | `submission.csv` with `id, satisfaction` |

### 🆕 What changed since Version 2

Before V3 took off, we sat down in the hangar with the five best-scoring public notebooks of this
competition, collected their ideas and **re-measured every one of them on our own folds**.

| | Version 2 | Version 3 |
|---|---|---|
| 🤖 Generator fingerprints | – | **digits** and **GPT-2 token keys** of distances, ages and delays |
| 📚 Original survey | 2 captain models | captains + the survey's satisfaction rate for every single value |
| 🎯 Target encoding | route and age | + route ÷ 10, route mod 1000, age × class, route × class, age × trip purpose |
| ⭐ Rating summaries | 9 row-level summaries | removed: they cost column-sampling slots and added nothing |
| 🟦 LightGBM | learning rate 0.03 | slower and steadier: 0.02, larger leaves |
| 🧠 Neural co-pilot | RealMLP, default-style settings | **RealMLP retuned** with the community's settings (+0.0009 AUC on one fold) |

## 🧳 Baggage claim: loading the data

The loader searches the Kaggle input folder first and a local `data/` folder second, so the same
notebook runs on Kaggle and on a laptop.

```python
DATA_ROOTS = ["/kaggle/input", "data"]
COMP = "playground-series-s6e10"


def find_file(name, hint):
    """First file called `name` whose path contains `hint`, searched under the data roots."""
    for root in DATA_ROOTS:
        hits = sorted(p for p in glob.glob(os.path.join(root, "**", name), recursive=True)
                      if hint in p.replace("\\", "/"))
        if hits:
            return hits[0]
    return None


def to_binary(s):
    """Map True/False, 1/0 or 'satisfied'/'neutral or dissatisfied' to 1/0."""
    if s.dtype == bool:
        return s.astype(int)
    return s.astype(str).str.strip().str.lower().isin(["true", "1", "1.0", "satisfied"]).astype(int)


paths = {f: find_file(f, COMP) for f in ("train.csv", "test.csv", "sample_submission.csv")}
assert all(paths.values()), f"Competition files not found under {DATA_ROOTS}: {paths}"
train = pd.read_csv(paths["train.csv"])
test = pd.read_csv(paths["test.csv"])
sample_sub = pd.read_csv(paths["sample_submission.csv"])

og_paths = [p for p in (find_file(f, "airline-passenger-satisfaction") for f in ("train.csv", "test.csv")) if p]
original = (pd.concat([pd.read_csv(p) for p in og_paths], ignore_index=True)
            if CFG.USE_ORIGINAL and og_paths else None)

if CFG.FAST_DEV:
    train = train.sample(CFG.DEV_ROWS, random_state=CFG.SEED).reset_index(drop=True)

y = to_binary(train[CFG.TARGET]).values
CATS = ["Gender", "Customer Type", "Type of Travel", "Class"]
NUMS = ["Age", "Flight Distance", "Departure Delay in Minutes", "Arrival Delay in Minutes"]
RATINGS = [c for c in test.columns if c not in CATS + NUMS + ["id"]]
BASE = NUMS + RATINGS + CATS

print(f"🧳 train {train.shape} · test {test.shape} · sample submission {sample_sub.shape}")
print(f"👨‍✈️ original survey: {'not attached' if original is None else f'{original.shape[0]:,} rows'}")
print(f"⭐ {len(RATINGS)} service ratings: {RATINGS}")
```

## 💺 Who is happy on board?

Three boarding-pass fields describe *why* and *how* someone flies: cabin **class**, trip
**purpose** and **loyalty**. Crossing them gives the clearest first picture of the passenger base.

```python
seg_df = train.assign(satisfied=y)
seg = seg_df.pivot_table(index="Class", columns=["Type of Travel", "Customer Type"],
                         values="satisfied", aggfunc="mean")
cnt = seg_df.pivot_table(index="Class", columns=["Type of Travel", "Customer Type"],
                         values="satisfied", aggfunc="size")
order = [c for c in ["Business", "Eco Plus", "Eco"] if c in seg.index]
seg, cnt = seg.loc[order], cnt.loc[order]

fig, ax = plt.subplots(figsize=(10, 3.8))
im = ax.imshow(seg.values, cmap=SEQ_BLUE, vmin=0, vmax=1, aspect="auto")
for i in range(seg.shape[0]):
    for j in range(seg.shape[1]):
        v, n = seg.values[i, j], cnt.values[i, j]
        if np.isnan(v):
            continue
        ax.text(j, i, f"{v:.0%}\n{int(n):,} pax", ha="center", va="center", fontsize=10,
                color="#ffffff" if v > 0.5 else INK)
ax.set_xticks(range(seg.shape[1]), [f"{a}\n{b}" for a, b in seg.columns])
ax.set_yticks(range(seg.shape[0]), seg.index)
ax.grid(False)
for s in ax.spines.values():
    s.set_visible(False)
cbar = fig.colorbar(im, ax=ax, format=mtick.PercentFormatter(1.0), fraction=0.03)
cbar.outline.set_visible(False)
ax.set_title("Satisfaction rate by cabin class, trip purpose and loyalty")
plt.tight_layout()
plt.show()

for col in ["Class", "Type of Travel", "Customer Type", "Gender"]:
    rates = seg_df.groupby(col)["satisfied"].mean().sort_values(ascending=False)
    print(f"{col:>15}: " + " | ".join(f"{k} {v:.1%}" for k, v in rates.items()))
```

## ⭐ The service touchpoints

Passengers rated 13 moments of the journey from 0 to 5 (plus the boarding-pass fields).
How well does **each feature on its own** rank happy passengers above unhappy ones? A single-feature
ROC AUC answers that directly: 0.5 is a coin flip, 1.0 is perfect separation.

```python
signal = []
for col in BASE:
    x = train[col]
    if col in CATS:                                   # order categories by their satisfaction rate
        x = x.map(pd.Series(y).groupby(x.values).mean())
    x = x.fillna(x.median())
    auc = roc_auc_score(y, x)
    signal.append((col, max(auc, 1 - auc), "higher → happier" if auc >= 0.5 else "higher → unhappier"))
signal = pd.DataFrame(signal, columns=["feature", "auc", "direction"]).sort_values("auc")

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(signal["feature"], signal["auc"] - 0.5, left=0.5, color=BLUE, height=0.62)
for _, r in signal.tail(5).iterrows():
    ax.text(r["auc"] + 0.004, r["feature"], f"{r['auc']:.3f}", va="center", color=INK_2, fontsize=9.5)
ax.set_xlim(0.5, min(1.0, signal["auc"].max() + 0.06))
ax.grid(axis="y", visible=False)
ax.set_xlabel("single-feature ROC AUC (0.5 = no signal)")
ax.set_title("How well each feature ranks satisfied passengers on its own")
plt.tight_layout()
plt.show()
```

```python
overall = y.mean()
levels = sorted(set().union(*[train[c].dropna().unique() for c in RATINGS]))
by_level = pd.DataFrame({c: pd.Series(y).groupby(train[c].values).mean() for c in RATINGS}).T
by_level = by_level.reindex(columns=levels)
by_level = by_level.loc[signal[signal["feature"].isin(RATINGS)].sort_values("auc", ascending=False)["feature"]]

fig, ax = plt.subplots(figsize=(10, 6.2))
norm = TwoSlopeNorm(vmin=0, vcenter=overall, vmax=1)
im = ax.imshow(np.ma.masked_invalid(by_level.values), cmap=DIVERGING, norm=norm, aspect="auto")
for i in range(by_level.shape[0]):
    for j in range(by_level.shape[1]):
        v = by_level.values[i, j]
        if not np.isnan(v):
            ax.text(j, i, f"{v:.0%}", ha="center", va="center", fontsize=9,
                    color="#ffffff" if abs(v - overall) > 0.3 else INK)
ax.set_xticks(range(len(levels)), [f"{int(l)}" for l in levels])
ax.set_yticks(range(len(by_level)), by_level.index)
ax.set_xlabel("rating given by the passenger")
ax.grid(False)
for s in ax.spines.values():
    s.set_visible(False)
cbar = fig.colorbar(im, ax=ax, format=mtick.PercentFormatter(1.0), fraction=0.03)
cbar.outline.set_visible(False)
ax.set_title(f"Satisfaction rate by rating level (grey = average {overall:.0%}, blue = above, red = below)")
plt.tight_layout()
plt.show()
```

## 🗺️ The hidden route ID

`Flight Distance` looks like a continuous number, but the synthetic generator re-used a limited set
of values from the original survey. If it really were a distance, nearby distances would have nearly
identical satisfaction rates. Let's check.

```python
FD = "Flight Distance"
all_fd = pd.concat([train[FD], test[FD]])
route = (pd.DataFrame({"fd": train[FD].values, "y": y})
         .groupby("fd")["y"].agg(rate="mean", n="size").reset_index())
route_big = route[route["n"] >= 100].sort_values("fd")

jump = route_big["rate"].diff().abs().median()                        # observed neighbour-to-neighbour jump
noise = np.sqrt(route_big["rate"] * (1 - route_big["rate"]) / route_big["n"]).median() * np.sqrt(2) * 0.674
tiles([
    ("🗺️", f"{all_fd.nunique():,}", "distinct flight distances (train + test)"),
    ("🧍", f"{len(all_fd) / all_fd.nunique():,.0f}", "passengers per distance on average"),
    ("📈", f"{jump:.1%}", "median jump between neighbouring distances"),
    ("🎲", f"{noise:.1%}", "jump expected from sampling noise alone"),
])

trend = (route_big.assign(bin=(route_big["fd"] // 250) * 250 + 125)
         .groupby("bin").apply(lambda g: np.average(g["rate"], weights=g["n"])))
fig, ax = plt.subplots(figsize=(11, 4.6))
ax.scatter(route_big["fd"], route_big["rate"], s=9, color=BLUE, alpha=0.45, linewidths=0,
           label="one flight distance (≥100 passengers)")
ax.plot(trend.index, trend.values, color=ORANGE, lw=2, label="smooth trend (250-mile bins)")
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
ax.set_xlabel("flight distance (miles)")
ax.set_ylabel("satisfaction rate")
ax.legend(loc="upper left")
ax.set_title("Neighbouring distances behave very differently: Flight Distance acts like a route ID")
plt.tight_layout()
plt.show()
```

<div style="border-left:5px solid #2a78d6; background:#f4f8fd; padding:12px 16px; border-radius:8px; margin:8px 0; color:#0b0b0b;">
<b>💡 Insight: treat Flight Distance as a category, not just a number</b><br>
There are only a few thousand distinct distances, each shared by hundreds of passengers, and neighbouring values jump far more than sampling noise allows. The value works like a hidden <b>route ID</b>. There is a real distance effect too: long-haul routes are satisfied about twice as often as short hops, but individual routes scatter widely around that trend. A tree with 255 histogram bins cannot isolate thousands of routes, so in the ground-crew chapter we describe each route explicitly: how often it appears, who flies it, and (inside the CV folds) how satisfied its passengers are. <i>Credit: this idea was surfaced by the Kaggle community early in the competition.</i>
</div>

## ⏱️ The delay paradox

Intuition says delays ruin a trip. They do hurt the passengers who suffer them, yet delay is one of the
weakest signals in the whole dataset. How can both be true?

```python
bins = [-1, 0, 15, 60, 180, np.inf]
labels = ["on time", "1–15 min", "16–60 min", "1–3 h", "3 h +"]
bucket = pd.cut(train["Departure Delay in Minutes"], bins=bins, labels=labels)
delay = pd.Series(y).groupby(bucket.values, observed=False).agg(rate="mean", n="size").reindex(labels)

fig, ax = plt.subplots(figsize=(9, 4))
ax.bar(delay.index.astype(str), delay["rate"], color=BLUE, width=0.6)
ax.axhline(overall, color=MUTED, lw=1)
ax.text(len(delay) - 0.5, overall + 0.01, f"overall {overall:.0%}", ha="right", color=INK_2, fontsize=9)
for i, (r, n) in enumerate(zip(delay["rate"], delay["n"])):
    if n > 0:
        ax.text(i, r + 0.01, f"{r:.0%}", ha="center", color=INK_2, fontsize=9.5)
        ax.text(i, 0.015, f"{int(n):,}", ha="center", color="#ffffff", fontsize=8.5)
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
ax.set_ylim(0, max(delay["rate"].max(), overall) + 0.1)
ax.grid(axis="x", visible=False)
ax.set_ylabel("satisfaction rate")
ax.set_title("Satisfaction rate by departure delay (number inside each bar = passengers)")
plt.tight_layout()
plt.show()

d, a = train["Departure Delay in Minutes"], train["Arrival Delay in Minutes"]
print(f"departure vs arrival delay correlation: {d.corr(a):.3f}  |  "
      f"share of passengers with no departure delay: {(d == 0).mean():.1%}")
```

<div style="border-left:5px solid #2a78d6; background:#f4f8fd; padding:12px 16px; border-radius:8px; margin:8px 0; color:#0b0b0b;">
<b>💡 Insight: delays hurt, but they are rare</b><br>
About <b>nine in ten</b> passengers depart on time. For the rest, satisfaction falls clearly as the delay grows, but because so few passengers are affected, each delay column ranks passengers barely better than a coin flip (single-feature AUC ≈ 0.51). That is the paradox. Departure and arrival delays are strongly correlated, so we keep both, add their sum and their difference (time made up in the air), and let the models decide how much to trust them.

## 🛠️ Ground crew: feature engineering

The ground crew prepares the V3 feature set. Everything except the target encoding is computed
on train + test together **without the competition target**, so it cannot leak labels.

| Group | Features | Why |
|---|---|---|
| ⏱️ Delay features | total delay, delay recovered in the air, missing-arrival flag | delays as one story, not two columns |
| 🗺️ Route profile | how often each Flight Distance appears + the average passenger on that route | describes the hidden route ID without the target |
| 🔢 Digits | distance `mod 10/100/1000`, `÷10`, `÷100`, length; age `mod 10` | the generator's digit fingerprint |
| 🤖 GPT-2 tokens | first token, last token and token count of distance, age and both delays | the generator's text fingerprint |
| 📚 Real-survey lookups | the real survey's satisfaction rate for each value of each column, and for trip × class × loyalty | real-world knowledge, no competition label involved |
| 🎯 Target encoding | out-of-fold satisfaction rate of routes, ages and their crosses | computed **inside** each CV fold with sklearn's cross-fitted `TargetEncoder` |

```python
CAT_MAPS = {c: {v: i for i, v in enumerate(sorted(pd.concat([train[c], test[c]]).dropna().unique()))}
            for c in CATS}


def encode_base(df):
    """Raw features with categories as stable integer codes (same codes for train, test, original)."""
    X = df[BASE].copy()
    for c in CATS:
        X[c] = X[c].map(CAT_MAPS[c]).fillna(-1).astype("int16")
    return X


ROUTE_PROFILE = ["Age", "Class", "Type of Travel", "Customer Type", "Online boarding", "Inflight wifi service",
                 "Inflight entertainment", "Seat comfort", "Ease of Online booking", "rating_mean",
                 "Departure Delay in Minutes"]
TOKEN_COLS = ["Flight Distance", "Age", "Departure Delay in Minutes", "Arrival Delay in Minutes"]


def gpt2_tokens(values):
    """{value: token ids} for every distinct value, written the way a text generator would see it."""
    return {v: GPT2.encode(" " + str(int(v))) for v in values.dropna().unique()}


def build_features(df):
    X = encode_base(df)
    dep, arr = X["Departure Delay in Minutes"], X["Arrival Delay in Minutes"]
    X["arrival_delay_missing"] = arr.isna().astype("int8")
    X["total_delay"] = dep + arr.fillna(dep)
    X["delay_recovered"] = dep - arr

    # 🗺️ the hidden route ID
    X["route_count"] = X[FD].map(X[FD].value_counts())
    profile = X.assign(rating_mean=X[RATINGS].mean(axis=1)).groupby(FD)[ROUTE_PROFILE].transform("mean")
    X[[f"route_avg_{c}" for c in ROUTE_PROFILE]] = profile.values

    # 🔢 digit fingerprints
    fd, age = df[FD], df["Age"]
    X["fd_mod10"], X["fd_mod100"], X["fd_mod1000"] = fd % 10, fd % 100, fd % 1000
    X["fd_div10"], X["fd_div100"], X["age_mod10"] = fd // 10, fd // 100, age % 10
    X["fd_length"] = fd.astype(int).astype(str).str.len()

    # 🤖 GPT-2 token fingerprints
    for c in TOKEN_COLS:
        toks = gpt2_tokens(df[c])
        X[f"{c} tok_first"] = df[c].map({v: t[0] for v, t in toks.items()})
        X[f"{c} tok_last"] = df[c].map({v: t[-1] + 100_000 * len(t) for v, t in toks.items()})
        X[f"{c} tok_count"] = df[c].map({v: len(t) for v, t in toks.items()})

    # 📚 what the real survey says about each value (no competition label involved)
    if original is not None:
        og_y = to_binary(original[CFG.TARGET])
        for c in BASE:
            X[f"{c} og_rate"] = df[c].map(og_y.groupby(original[c].values).mean())
        trio = ["Type of Travel", "Class", "Customer Type"]
        rate = original[trio].assign(y=og_y.values).groupby(trio)["y"].mean().rename("trio_og_rate")
        X["trio_og_rate"] = df[trio].merge(rate, left_on=trio, right_index=True, how="left")["trio_og_rate"].values

    # 🎯 keys for the fold-safe target encoding (crosses of age / route with class and trip purpose)
    X["key_age_class"] = age * 10 + X["Class"]
    X["key_fd_class"] = fd * 10 + X["Class"]
    X["key_age_travel"] = age * 10 + X["Type of Travel"]

    X.columns = [clean(c) for c in X.columns]
    return X.astype("float32")


n_train = len(train)
full = pd.concat([train[BASE], test[BASE]], ignore_index=True)
FEATURES = build_features(full)
print(f"🛠️ feature matrix: {FEATURES.shape[0]:,} rows × {FEATURES.shape[1]} features")
display(FEATURES.head(3))
```

## 👨‍✈️ The veteran captains: learning from the original survey

The competition data was synthesised from a real survey of ~130k passengers. Models trained
**only on that original survey** have never seen a single competition label, so their predictions
for the competition passengers are leak-free features: the opinion of veteran captains who flew
these routes with real passengers before.

Two captains with different flying styles are on duty: a LightGBM and a CatBoost model. In version 2
the second captain added **+0.00013 AUC**, and a third one added nothing more, so the crew stays at two.

```python
TEACHER_NAMES = []
if original is not None:
    og = original.dropna(subset=[c for c in BASE if c != "Arrival Delay in Minutes"])
    og = og[BASE + [CFG.TARGET]].drop_duplicates()
    og_y = to_binary(og[CFG.TARGET]).values
    X_og = encode_base(og).rename(columns=clean)
    X_all = encode_base(full).rename(columns=clean)

    captains = {
        "og_lgb": lgb.LGBMClassifier(n_estimators=700, learning_rate=0.03, num_leaves=63, subsample=0.8,
                                     subsample_freq=1, colsample_bytree=0.7, random_state=CFG.SEED, verbose=-1),
        "og_cat": CatBoostClassifier(iterations=1500, learning_rate=0.05, depth=8,
                                     cat_features=[clean(c) for c in CATS], random_seed=CFG.SEED, verbose=0,
                                     allow_writing_files=False, task_type="GPU" if CFG.GPU else "CPU"),
    }
    report = []
    for name, model in captains.items():
        t0 = time.time()
        model.fit(X_og, og_y)
        FEATURES[name] = model.predict_proba(X_all)[:, 1].astype("float32")
        TEACHER_NAMES.append(name)
        report.append((name, roc_auc_score(y, FEATURES[name].values[:n_train]), time.time() - t0))
    tiles([("👨‍✈️", f"{len(og):,}", "original survey passengers")]
          + [("🎯", f"{auc:.4f}", f"{name}: AUC on competition train ({sec:.0f}s, zero competition labels)")
             for name, auc, sec in report])
    del X_og, X_all
else:
    print("👨‍✈️ Original survey not attached: flying without the veteran captains.")
```

### 🌪️ The turbulence floor

How far can the captains be trusted? Group the competition passengers by the captains' average
predicted probability and compare it with the share that is actually labelled *satisfied*.

<b>💡 Insight: why nobody scores 0.99</b><br>
Two things stand out. First, the curve never reaches 0% or 100%: even among passengers the captains consider certain, a few percent carry the opposite label. That is label noise from the synthetic generator, a <b>turbulence floor</b> no model can fly below, and the reason the leaderboard top sits near 0.961. ## 🛫 Takeoff: three pilots, five folds

Every model flies the **same 5 stratified folds**, so their out-of-fold (OOF) predictions line up row
for row and can be stacked honestly later.

Inside each fold the **target encodings** (route, age, route ÷ 10, route mod 1000 and the age / route
crosses) are fitted on the training part only, with an inner cross-fit, then applied to the validation
fold and the test set.

| Pilot | Style |
|---|---|
| 🟦 LightGBM | leaf-wise trees with 1023 bins so it can resolve individual routes; slower learning rate in V3 |
| 🟧 XGBoost | depth-wise trees; GPU if available |
| 🐱 CatBoost | ordered boosting with native categorical handling, including the route ID itself |

```python
TE_COLS = [clean(FD), "age", "fd_div10", "fd_mod1000", "key_age_class", "key_fd_class", "key_age_travel"]
TE_NAMES = [f"te_{c}" for c in TE_COLS]
folds = list(StratifiedKFold(CFG.N_FOLDS, shuffle=True, random_state=CFG.SEED).split(X_train, y))


def fold_data(k):
    """Train / valid / test matrices for fold k, with fold-safe target encodings added."""
    tr_idx, va_idx = folds[k]
    X_tr, X_va, X_te = X_train.iloc[tr_idx].copy(), X_train.iloc[va_idx].copy(), X_test.copy()
    enc = TargetEncoder(target_type="binary", cv=5, shuffle=True, random_state=CFG.SEED)
    X_tr[TE_NAMES] = enc.fit_transform(X_tr[TE_COLS], y[tr_idx]).astype("float32")
    X_va[TE_NAMES] = enc.transform(X_va[TE_COLS]).astype("float32")
    X_te[TE_NAMES] = enc.transform(X_te[TE_COLS]).astype("float32")
    return X_tr, y[tr_idx], X_va, y[va_idx], X_te


def run_cv(name, fit_predict, unit="trees"):
    """Train one model type on every fold; returns OOF + averaged test predictions."""
    oof, test_pred, scores, importances = np.zeros(len(X_train)), np.zeros(len(X_test)), [], []
    t0 = time.time()
    print(f"🛫 {name} is taking off")
    for k in range(CFG.N_FOLDS):
        X_tr, y_tr, X_va, y_va, X_te = fold_data(k)
        p_va, p_te, imp, size = fit_predict(X_tr, y_tr, X_va, y_va, X_te)
        oof[folds[k][1]] = p_va
        test_pred += p_te / CFG.N_FOLDS
        scores.append(roc_auc_score(y_va, p_va))
        if imp is not None:
            importances.append(imp)
        print(f"   ✈️ fold {k + 1}/{CFG.N_FOLDS} · AUC {scores[-1]:.5f} · {size} {unit} · {time.time() - t0:,.0f}s")
        del X_tr, X_va, X_te; gc.collect()
    score = roc_auc_score(y, oof)
    print(f"🏁 {name}: OOF AUC {score:.5f}  (fold mean {np.mean(scores):.5f} ± {np.std(scores):.5f})\n")
    return dict(name=name, oof=oof, test=test_pred, score=score, folds=scores, seconds=time.time() - t0,
                importance=pd.concat(importances, axis=1).mean(axis=1) if importances else None)


results = {}
```

```python
LGB_PARAMS = dict(
    objective="binary", learning_rate=0.02, n_estimators=20_000, num_leaves=127, min_child_samples=100,
    colsample_bytree=0.5, subsample=0.8, subsample_freq=1, reg_lambda=1.0, max_bin=1023,
    random_state=CFG.SEED, n_jobs=-1, verbose=-1,
)


def lgb_fit_predict(X_tr, y_tr, X_va, y_va, X_te):
    model = lgb.LGBMClassifier(**LGB_PARAMS)
    model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], eval_metric="auc",
              callbacks=[lgb.early_stopping(CFG.EARLY_STOP, verbose=False)])
    imp = pd.Series(model.booster_.feature_importance("gain"), index=X_tr.columns)
    return model.predict_proba(X_va)[:, 1], model.predict_proba(X_te)[:, 1], imp, model.best_iteration_


results["LightGBM"] = run_cv("🟦 LightGBM", lgb_fit_predict)
```

```python
XGB_PARAMS = dict(
    n_estimators=10_000, learning_rate=0.03, max_depth=7, min_child_weight=5, subsample=0.8,
    colsample_bytree=0.5, reg_lambda=5.0, max_bin=1024, tree_method="hist",
    device="cuda" if CFG.GPU else "cpu", eval_metric="auc", early_stopping_rounds=CFG.EARLY_STOP,
    random_state=CFG.SEED, n_jobs=-1,
)


def xgb_fit_predict(X_tr, y_tr, X_va, y_va, X_te):
    model = xgb.XGBClassifier(**XGB_PARAMS)
    model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], verbose=False)
    best = (0, model.best_iteration + 1)
    return (model.predict_proba(X_va, iteration_range=best)[:, 1],
            model.predict_proba(X_te, iteration_range=best)[:, 1], None, model.best_iteration + 1)


results["XGBoost"] = run_cv("🟧 XGBoost", xgb_fit_predict)
```

```python
CB_CATS = [clean(c) for c in CATS] + ["route_id"]
CB_PARAMS = dict(
    iterations=6_000, learning_rate=0.08, depth=7, l2_leaf_reg=5, eval_metric="AUC",
    od_type="Iter", od_wait=CFG.EARLY_STOP, random_seed=CFG.SEED, verbose=0,
    task_type="GPU" if CFG.GPU else "CPU", thread_count=-1, allow_writing_files=False,
)


def cb_frame(X):
    """CatBoost wants integer categories; the raw route ID becomes an extra categorical column."""
    X = X.assign(route_id=X[clean(FD)])
    X[CB_CATS] = X[CB_CATS].astype("int32")
    return X


def cb_fit_predict(X_tr, y_tr, X_va, y_va, X_te):
    X_tr, X_va, X_te = cb_frame(X_tr), cb_frame(X_va), cb_frame(X_te)
    model = CatBoostClassifier(**CB_PARAMS)
    model.fit(Pool(X_tr, y_tr, cat_features=CB_CATS), eval_set=Pool(X_va, y_va, cat_features=CB_CATS),
              use_best_model=True)
    return (model.predict_proba(X_va)[:, 1], model.predict_proba(X_te)[:, 1], None,
            model.get_best_iteration() + 1)


results["CatBoost"] = run_cv("🐱 CatBoost", cb_fit_predict)
```

## 🧠 The neural co-pilot, retuned

Trees cut the feature space into boxes; a neural network draws smooth surfaces. Their mistakes differ,
and different mistakes are exactly what a stack can exploit. The co-pilot is **RealMLP**
(Holzmüller et al., 2024, from the `pytabkit` package).

In version 2 it was the weakest member (0.96010 OOF). The hangar talk pointed at the reason: its settings.
We tried our own tweaks first, then the settings the community tuned for this competition, all on fold 0:

| Co-pilot settings (fold 0) | AUC | Δ |
|---|---:|---:|
| V2 co-pilot: 4 networks × 4 epochs, default-style settings | 0.96025 | – |
| + 6 epochs instead of 4 | 0.96026 | +0.00001 |
| + wider layers (512-256-128) | 0.96021 | −0.00004 |
| + light label smoothing (0.01) | 0.96028 | +0.00003 |
| community settings (learning-rate and weight-decay schedules, smaller embeddings, wider one-hot, PLR settings, L2-normalised inputs) | 0.96114 | +0.00089 |
| community settings + the V3 feature set | 0.96130 | +0.00105 |
| **… with 8 networks side by side instead of 4** (what flies in V3) | **0.96140** | **+0.00115** |

No single tweak helped; the whole tuned recipe did, and the hangar-talk features help the co-pilot as well. What we keep from version 2: route, age and every
rating also enter as **categories** (one embedding per route, the hidden route ID), and the captains'
predictions enter as **logits**.

```python
NN_TWINS = {"route_id": clean(FD), "age_cat": "age", **{f"{clean(r)}_cat": clean(r) for r in RATINGS}}
NN_CATEGORIES = {c: sorted(FEATURES[c].astype(int).unique()) for c in [clean(c) for c in CATS]}
NN_CATEGORIES.update({twin: sorted(FEATURES[src].astype(int).unique()) for twin, src in NN_TWINS.items()})
NN_CAT_COLS = list(NN_CATEGORIES)

REALMLP_ENSEMBLE = 8                                   # networks trained side by side inside one model
REALMLP_PARAMS = dict(
    n_epochs=1 if CFG.FAST_DEV else 4, n_ens=2 if CFG.FAST_DEV else REALMLP_ENSEMBLE, batch_size=256,
    use_early_stopping=False, val_metric_name="1-auc_ovr",
    device="cuda" if CFG.GPU else "cpu", random_state=CFG.SEED, verbosity=0,
    # community settings for this competition: Vladimir Demidov's RealMLP notebook, as adapted by goodpjw2008
    lr=0.053, wd=0.0236, sq_mom=0.988, lr_sched="lin_cos_log_15", wd_sched="cos_log_15",
    first_layer_lr_factor=0.25, embedding_size=5, max_one_hot_cat_size=18,
    hidden_sizes=[512, 256, 128], act="silu", p_drop=0.05, p_drop_sched="expm4t",
    plr_hidden_1=16, plr_hidden_2=8, plr_act_name="gelu", plr_lr_factor=0.1151, plr_sigma=2.33,
    use_ls=True, ls_eps=0.01, ls_eps_sched="sqrt_cos", add_front_scale=False,
    bias_init_mode="neg-uniform-dynamic-2",
    tfms=["one_hot", "median_center", "robust_scale", "smooth_clip", "embedding", "l2_normalize"],
)


def nn_frame(X):
    """RealMLP input: captain logits, categorical twins with fixed category lists, no gaps."""
    X = X.copy()
    for c in TEACHER_NAMES:
        q = X[c].clip(1e-5, 1 - 1e-5)
        X[c] = np.log(q / (1 - q))
    for twin, src in NN_TWINS.items():
        X[twin] = X[src]
    X = X.fillna(0)                                     # only the arrival-delay columns have gaps
    for c in NN_CAT_COLS:
        X[c] = pd.Categorical(X[c].astype(int), categories=NN_CATEGORIES[c])
    return X


def realmlp_fit_predict(X_tr, y_tr, X_va, y_va, X_te):
    X_va_nn = nn_frame(X_va)
    model = RealMLP_TD_Classifier(**REALMLP_PARAMS)
    model.fit(nn_frame(X_tr), y_tr, X_val=X_va_nn, y_val=y_va, cat_col_names=NN_CAT_COLS)
    p_va, p_te = model.predict_proba(X_va_nn)[:, 1], model.predict_proba(nn_frame(X_te))[:, 1]
    del model; gc.collect(); torch.cuda.empty_cache()
    return p_va, p_te, None, REALMLP_PARAMS["n_epochs"]


if RUN_REALMLP:
    results["RealMLP"] = run_cv("🧠 RealMLP", realmlp_fit_predict, unit="epochs")
else:
    print("🧠 The co-pilot stays on the ground this time (pytabkit is not available).")
```

## 🏗️ Control tower: stacking the pilots

The control tower listens to every pilot and both captains, and decides how much to trust each.
Technically: a **logistic regression on the logits** of the out-of-fold predictions. To keep the score
honest, the stack is itself evaluated out of fold: for each fold, the stacker is trained on the other
four folds' OOF predictions only. The simple average of the pilots is shown as a reference, and the
better of the two lands the plane.

```python
def logit(p):
    p = np.clip(np.asarray(p, dtype="float64"), 1e-6, 1 - 1e-6)
    return np.log(p / (1 - p))


names = list(results)
inputs = names + TEACHER_NAMES
Z = np.column_stack([logit(results[n]["oof"]) for n in names] + [logit(X_train[t]) for t in TEACHER_NAMES])
Z_test = np.column_stack([logit(results[n]["test"]) for n in names] + [logit(X_test[t]) for t in TEACHER_NAMES])

stack_oof = np.zeros(len(y))
for tr_idx, va_idx in folds:
    stacker = LogisticRegression(C=1.0, max_iter=2000).fit(Z[tr_idx], y[tr_idx])
    stack_oof[va_idx] = stacker.predict_proba(Z[va_idx])[:, 1]
stacker = LogisticRegression(C=1.0, max_iter=2000).fit(Z, y)
stack_test = stacker.predict_proba(Z_test)[:, 1]

mean_oof = np.mean([results[n]["oof"] for n in names], axis=0)
mean_test = np.mean([results[n]["test"] for n in names], axis=0)
scores = {n: results[n]["score"] for n in names}
scores["Simple average"] = roc_auc_score(y, mean_oof)
scores["Stack"] = roc_auc_score(y, stack_oof)
use_stack = scores["Stack"] >= scores["Simple average"]
final_name = "Stack" if use_stack else "Simple average"
final_oof, final_test = (stack_oof, stack_test) if use_stack else (mean_oof, mean_test)
best_single = max(results[n]["score"] for n in names)

board = pd.DataFrame({"OOF AUC": scores})
board["vs best single"] = board["OOF AUC"] - best_single
board["minutes"] = [results[n]["seconds"] / 60 if n in results else np.nan for n in board.index]
display(board.style.format({"OOF AUC": "{:.5f}", "vs best single": "{:+.5f}", "minutes": "{:.1f}"}, na_rep="–"))
display(pd.Series(stacker.coef_[0], index=inputs, name="stack weight (on logits)").to_frame()
        .style.format("{:.3f}"))

order = board["OOF AUC"].sort_values()
fig, ax = plt.subplots(figsize=(9, 0.55 * len(order) + 1.2))
colors = [ORANGE if n == final_name else BLUE for n in order.index]
ax.hlines(order.index, order.min() - 0.0003, order.values, color=colors, lw=2)
ax.scatter(order.values, order.index, color=colors, s=60, zorder=3)
for n, v in order.items():
    ax.text(v + 0.00003, n, f"{v:.5f}", va="center", color=INK_2, fontsize=9.5)
ax.set_xlim(order.min() - 0.0003, order.max() + 0.0003)
ax.grid(axis="y", visible=False)
ax.set_xlabel("out-of-fold ROC AUC (axis zoomed in)")
ax.set_title("Every pilot, and the formation that flies the submission (orange)")
plt.tight_layout()
plt.show()
```
