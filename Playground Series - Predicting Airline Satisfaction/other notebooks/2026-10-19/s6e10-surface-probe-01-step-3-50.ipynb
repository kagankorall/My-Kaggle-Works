# S6E10 Empirical Surface Probe 01

Fitted-gradient step: **3.50**. The local 96-vector response surface predicts this probe near 0.96205; only Kaggle scoring can verify it.

This notebook writes `submission.csv` only and never calls the competition submission API.

```python
from pathlib import Path
import numpy as np
import pandas as pd

STEP = 3.50
anchor_hits = list(Path('/kaggle/input').rglob('nina_rank_anchor.csv'))
direction_hits = list(Path('/kaggle/input').rglob('fitted_ascent_direction.npy'))
assert len(anchor_hits) == 1, anchor_hits
assert len(direction_hits) == 1, direction_hits
anchor = pd.read_csv(anchor_hits[0])
direction = np.load(direction_hits[0])
assert list(anchor.columns) == ['id', 'satisfaction'] and len(anchor) == 299_844
assert len(direction) == len(anchor)

raw = anchor.satisfaction.to_numpy() + STEP * direction
pred = (pd.Series(raw).rank(method='average').to_numpy() - 0.5) / len(raw)
submission = pd.DataFrame({'id': anchor.id, 'satisfaction': pred})
assert submission.id.is_unique and submission.satisfaction.notna().all()
assert submission.satisfaction.between(0, 1).all()
submission.to_csv('submission.csv', index=False, float_format='%.9f')
print(f'step={STEP:.2f} rows={len(submission)} corr_anchor={np.corrcoef(pred, anchor.satisfaction)[0,1]:.9f}')
submission.head()
```
