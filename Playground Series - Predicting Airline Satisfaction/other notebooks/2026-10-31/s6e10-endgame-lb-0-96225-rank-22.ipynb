`submission.csv` = the public tkms0816 0.96223 file + 1.75 × (Mirror Nudge Climb III − Mirror Pair Climb), in rank space → **public LB 0.96225**.

## How DO YOU NEED LESS? got us here

DO YOU NEED LESS? found its gains by moving a strong file a small step *away from* a weaker one (`1.06·meta − 0.05·XGBoost − 0.01·TabPFN` → 0.96218).
The same probe tells you something useful about any new anchor: if pushing it away from a model no longer helps, that direction is **already inside** the file and you should look for a different one.

That is exactly what happened on the new best public file.

## Step 1: test the old direction on the new anchor

The anchor `X` is tkms0816's public 0.96223 file. It already pushes goodpjw2008's meta-stack away from ten weaker files (including a TabPFN and an XGBoost) and adds Mirror Pair Climb's direction.

| file | public LB |
|---|---|
| `X` (tkms0816) | 0.96223 |
| 1.05·X − 0.05·XGBoost | 0.96222 |
| 1.06·X − 0.05·XGBoost − 0.01·TabPFN-3.5 | 0.96222 |

Both lost a point, so the XGBoost/TabPFN push is used up. We need a direction `X` doesn't contain yet.

## Step 2: find a direction X doesn't have

mayankraaz published several related files. Mirror Pair Climb (0.96218) is already inside `X`. Mirror Nudge Climb III (0.96219) is built the same way but with its own set of mirror nudges.
Their difference in rank space is what III has that Pair Climb doesn't:

`d = rank(Mirror Nudge Climb III) − rank(Mirror Pair Climb)`

`X` doesn't contain `d`, so adding it can still move the score. The blend is plain rank arithmetic, using the same `blend()` helper as DO YOU NEED LESS?:

`final = rank(X) + k · d`

## Step 3: walk the step size, one probe each

| k | public LB |
|---|---|
| 0.5 | 0.96224 |
| 1.0 | 0.96224 |
| 1.5 | 0.96225 |
| **1.75** | **0.96225** ← `submission.csv` |
| 2.0 | 0.96225 |
| 3.0 | 0.96223 |
| 1.5 + 0.5 × (III − Nudge Climb II) | 0.96224 |

The score rises up to about k = 1.5, stays flat until 2.0 and drops by 3.0. `k = 1.75` sits in the middle of the flat part.
Adding the Nudge Climb II difference on top lost a point. It points mostly the same way as `d` (cosine 0.77), so it just over-steps.

## Caveat

`d` comes from row-block nudges that were chosen on the public LB, and `k` was chosen there too. The public LB is only 20% of the test set, so on the private split this part may do nothing or cost a little.
If you pick this as a final submission, pair it with one CV-validated file.

Every probe file is written to the output folder (`blend_manifest.csv` lists the recipe for each), so you can re-submit any row above.
